```python
import torch
from torch import autograd
import numpy as np
import matplotlib.pyplot as plt
import torch.optim as optim
```

```python
x=torch.tensor([1.,2.,3.], requires_grad=True)
y = x ** 2
```

```python
outgrad=autograd.grad(y, x, torch.ones_like(y), retain_graph=True)
```

```python
outgrad
```

```python
outgrad=autograd.grad(y, x, torch.zeros_like(y), retain_graph=True)
outgrad
```

```python
outgrad=autograd.grad(y, x, torch.tensor([0.,1.,0.]), retain_graph=True)
outgrad
```

## 1. Краткая теория

### 1.1. Задача оптимизации

Пусть у нас есть функция потерь $L(w)$, которую хотим минимизировать по параметрам $w$. В линейной регрессии:

$$L(w, b) = \frac{1}{n}\sum_{i=1}^{n} (w x_i + b - y_i)^2$$

### 1.2. Градиентный спуск (Batch GD)

На каждом шаге используем **весь** датасет, чтобы посчитать градиент:

$$w_{t+1} = w_t - \eta \cdot \nabla L(w_t), \qquad \nabla L(w_t) = \frac{1}{n}\sum_{i=1}^{n}\nabla \ell_i(w_t)$$

где $\eta$ — learning rate (шаг обучения).

- Точный градиент, стабильное снижение лосса
- НО: Очень дорого на больших данных (нужен проход по всему датасету на каждом шаге)

### 1.3. Стохастический градиентный спуск (SGD)

На каждом шаге берём **один** случайный объект:

$$w_{t+1} = w_t - \eta \cdot \nabla \ell_i(w_t)$$

- Очень дёшево, позволяет «вылезать» из локальных минимумов за счёт шума
- НО: Сильные колебания, лосс «прыгает» — усреднённая траектория всё равно сходится

### 1.4. Mini-batch GD

Компромисс: на каждом шаге берём **мини-батч** из $B$ объектов:

$$w_{t+1} = w_t - \eta \cdot \frac{1}{B}\sum_{i \in \text{batch}}\nabla \ell_i(w_t)$$

- Устойчивее SGD, быстрее GD — именно так обучают нейросети
- НО: Появляется гиперпараметр $B$ (обычно 32–512)

## 2. Подготовка данных

Сгенерируем синтетический датасет для линейной регрессии: $y = 2x + 1 + \text{шум}$. Задача — восстановить коэффициенты $w$ и $b$ градиентным спуском.

````python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# --- Данные ---
n = 200
X = np.random.uniform(-3, 3, n)                     # признак
y = 2.0 * X + 1.0 + np.random.normal(0, 0.5, n)     # истинные w=2, b=1 + шум

# --- Визуализация ---
plt.figure(figsize=(6, 4))
plt.scatter(X, y, s=15, alpha=0.6, label='данные')
plt.plot(np.sort(X), 2*np.sort(X) + 1, 'r-', label='истинная прямая')
plt.xlabel('x'); plt.ylabel('y'); plt.legend(); plt.grid(alpha=0.3)
plt.show()
````

## 3. Задание 1: реализуйте функцию потерь и градиент

Для среднеквадратичной ошибки (MSE) с моделью $\hat{y} = wx + b$:

$$L = \frac{1}{n}\sum_{i}(wx_i + b - y_i)^2$$

$$\frac{\partial L}{\partial w} = \frac{2}{n}\sum_{i}(wx_i + b - y_i)\,x_i, \qquad \frac{\partial L}{\partial b} = \frac{2}{n}\sum_{i}(wx_i + b - y_i)$$

Реализуйте:
- `mse(y_pred, y)` — среднеквадратичная ошибка,
- `grads(X, y, w, b)` — градиенты по `w` и `b` (на всём переданном массиве — это пригодится и для батчей).

````python
# TODO 1: функция потерь и градиенты

def mse(y_pred, y):
    """Среднеквадратичная ошибка."""
    L=1/len(y_pred)*sum((y_pred-y)**2)
    return L
def grads(X, y, w, b):
    """Градиенты dL/dw и dL/db по переданной выборке (батчу)."""
    dL_dw=2*np.mean((w*X+b-y)*X)
    dL_db=2*np.mean((w*X+b-y))
    return dL_dw, dL_db
````

## 4. Задание 2: Batch Gradient Descent

Реализуйте классический GD:
1. Инициализируйте `w`, `b` нулями.
2. На каждой итерации посчитайте градиент **по всему датасету** и обновите параметры.
3. Записывайте значение лосса после каждой итерации в список `history`.
4. Верните `w, b, history`.

````python
# TODO 2: batch gradient descent

def gradient_descent(X, y, lr=0.1, n_iter=200):
    w, b = 0.0, 0.0
    history = []
    for _ in range(n_iter):
        Lw, Lb = grads(X, y, w, b)
        w=w-lr*Lw
        b=b-lr*Lb
        y_pred = w*X+b
        history.append(mse(y_pred,y))

    return w, b, history

w_gd, b_gd, hist_gd = gradient_descent(X, y, lr=0.1, n_iter=200)
print(f'GD: w = {w_gd:.3f}, b = {b_gd:.3f} (истина: w=2, b=1)')

plt.figure(figsize=(6, 4))
plt.plot(hist_gd)
plt.xlabel('итерация'); plt.ylabel('MSE'); plt.title('Сходимость Batch GD'); plt.grid(alpha=0.3)
plt.show()
````

## 5. Задание 3: Stochastic Gradient Descent

Теперь на каждой итерации используем **один** случайно выбранный объект.

Подсказки:
- `np.random.randint(n)` — случайный индекс;
- передавайте в `grads` массивы длины 1: `X[i:i+1]`, `y[i:i+1]`.

````python
# TODO 3: stochastic gradient descent

def sgd(X, y, lr=0.05, n_iter=200):
    n = len(X)
    w, b = 0.0, 0.0
    history = []
    for _ in range(n_iter):
        i = np.random.randint(n)
        Lw, Lb = grads(X[i:i+1], y[i:i+1], w, b)
        w=w-lr*Lw
        b=b-lr*Lb
        y_pred = w*X+b
        history.append(mse(y_pred,y))
    return w, b, history

w_sgd, b_sgd, hist_sgd = sgd(X, y, lr=0.05, n_iter=200)
print(f'SGD: w = {w_sgd:.3f}, b = {b_sgd:.3f}')

plt.figure(figsize=(6, 4))
plt.plot(hist_sgd)
plt.xlabel('итерация'); plt.ylabel('MSE'); plt.title('Сходимость SGD '); plt.grid(alpha=0.3)
plt.show()
````

## 6. Задание 4: Mini-batch Gradient Descent

Теперь берём батчи по `batch_size` объектов. За одну **эпоху** проходим по всему датасету.

Подсказки:
- перемешайте индексы `np.random.permutation(n)`;
- берите срез размером `batch_size`;
- `epochs` ниже = число эпох. Записывайте лосс после каждой эпохи.

````python
# TODO 4: mini-batch gradient descent

def minibatch_gd(X, y, lr=0.05, batch_size=32, epochs=20):
    n = len(X)
    w, b = 0.0, 0.0
    history = []
    for i in range(epochs):
        j = np.random.permutation(n)
        dL_dw, dL_db = grads(X[j][0:batch_size], y[j][0:batch_size], w, b)
        w = w - lr*dL_dw
        b = b - lr*dL_db
        y_pred = w*X[j][0:batch_size] + b
        L = mse(y_pred, y[j][0:batch_size])
        history.append(L)
    return w, b, history

w_mb, b_mb, hist_mb = minibatch_gd(X, y, lr=0.05, batch_size=32, epochs=20)
print(f'Mini-batch: w = {w_mb:.3f}, b = {b_mb:.3f}')

plt.figure(figsize=(6, 4))
plt.plot(hist_mb)
plt.xlabel('эпоха'); plt.ylabel('MSE'); plt.title('Сходимость Mini-batch GD'); plt.grid(alpha=0.3)
plt.show()
````

## 7. Сравнение методов

Чтобы сравнение было честным, посчитаем количество **обработанных объектов** по горизонтали: у GD одна итерация = $n$ объектов, у SGD одна итерация = 1 объект.

````python
n = len(X)

xs_gd  = [(i + 1) * n for i in range(len(hist_gd))]
xs_sgd = list(range(1, len(hist_sgd) + 1))
xs_mb  = [(i + 1) * n for i in range(len(hist_mb))]

plt.figure(figsize=(8, 5))
plt.plot(xs_gd, hist_gd, label='Batch GD', linewidth=2)
plt.plot(xs_sgd, hist_sgd, label='SGD', alpha=0.5)
plt.plot(xs_mb, hist_mb, label='Mini-batch (32)', linewidth=2)
plt.xscale('log')
plt.xlabel('обработано объектов (log)'); plt.ylabel('MSE')
plt.title('GD vs SGD vs Mini-batch'); plt.legend(); plt.grid(alpha=0.3)
plt.show()
````
